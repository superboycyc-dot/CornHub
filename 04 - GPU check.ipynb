# Keeboh GPU smoke test

This notebook verifies the runtime identity, writable workspace, and NVIDIA GPU visibility.

```python
import os
import platform
from pathlib import Path

print({
    'python': platform.python_version(),
    'workspace': str(Path.cwd()),
    'team_id_present': bool(os.getenv('KEEBOH_TEAM_ID')),
    'workspace_id_present': bool(os.getenv('KEEBOH_WORKSPACE_ID')),
})
```

```python
workspace_check = Path('/workspace/.keeboh-write-check')
workspace_check.write_text('ok\n', encoding='utf-8')
assert workspace_check.read_text(encoding='utf-8') == 'ok\n'
workspace_check.unlink()
print('/workspace is writable')
```

```python
import subprocess

result = subprocess.run(
    ['nvidia-smi'],
    check=False,
    capture_output=True,
    text=True,
)
print(result.stdout or result.stderr)
assert result.returncode == 0, 'nvidia-smi did not detect an NVIDIA GPU'
```
